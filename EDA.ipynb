```python
#Time series analysis
```

```python
df["Order Date"] = pd.to_datetime(
    df["Order Date"],
    errors="coerce"
)
```

```python
df["Month"] = df["Order Date"].dt.month_name()
df["Month Number"] = df["Order Date"].dt.month
df["Year"] = df["Order Date"].dt.year

```

```python
df[["Order Date", "Month", "Month Number", "Year"]].head()
```

```python
monthly_sales = (
    df.groupby(["Month Number", "Month"])["Sales"]
      .sum()
      .reset_index()
      .sort_values("Month Number")
)

monthly_sales
```

```python
plt.figure(figsize=(10,5))

plt.plot(monthly_sales["Month"],
         monthly_sales["Sales"],
         marker="o")

plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.xticks(rotation=45)

plt.show()
```

```python
#Sales usually faculates betweeen months but December recorded the most sales and feburary recorded the least sale
```

```python
#
```

```python
monthly_profit = (
    df.groupby(["Month Number", "Month"])["Gross Profit"]
      .sum()
      .reset_index()
      .sort_values("Month Number")
)

monthly_profit
```

```python
plt.figure(figsize=(10,5))

plt.plot(
    monthly_profit["Month"],
    monthly_profit["Gross Profit"],
    marker="o"
)

plt.title("Monthly Gross Profit")
plt.xlabel("Month")
plt.ylabel("Gross Profit")
plt.xticks(rotation=45)

plt.show()
```

```python
#December records the most profit followed by september and february records the least profit on of the reason might be less monthly sales in februray 
```

```python
monthly_orders = (
    df.groupby(["Month Number", "Month"])["Order ID"]
      .count()
      .reset_index()
      .sort_values("Month Number")
)

monthly_orders
```

```python
plt.figure(figsize=(10,5))

plt.plot(
    monthly_orders["Month"],
    monthly_orders["Order ID"],
    marker="o"
)

plt.title("Monthly Orders")
plt.xlabel("Month")
plt.ylabel("Number of Orders")
plt.xticks(rotation=45)

plt.show()
```

```python
#December records the most orders followed by September and februry records the least ammout of salaes
```

```python
#Top Customer
```

```python
top_customers = (
    df.groupby("Customer ID")["Sales"]
      .sum()
      .sort_values(ascending=False)
      .head(10)
)

top_customers
```
